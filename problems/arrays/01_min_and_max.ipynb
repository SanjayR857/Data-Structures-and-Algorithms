### Problem Statement
Given an array `arr` of size `N`, find the minimum and maximum elements in the array in minimum comparisons.

### Constraints
- `1 <= N <= 10^5`
- `1 <= arr[i] <= 10^12`

### 💡 Intuition & Approach
- **Approach 1 (Standard Linear Scan):** Initialize `min_val` and `max_val` to `arr[0]`. Iterate through the array from index `1` to `N-1`. Update `min_val = min(min_val, x)` and `max_val = max(max_val, x)`.
  - **Total Comparisons:** $2(n - 1)$
  - **Time Complexity:** $O(n)$
  - **Space Complexity:** $O(1)$
- **Approach 2 (Pairwise Tournament Optimization):** Compare elements in pairs `(arr[i], arr[i+1])`, then compare the smaller with `min_val` and larger with `max_val`.
  - **Total Comparisons:** $3n/2 - 2$ (reduces comparisons by 25%).

```python
def get_min_max(arr: list[int]) -> tuple[int, int]:
    """Returns (min_val, max_val) of an array in O(n) time and O(1) space."""
    if not arr:
        raise ValueError("Array cannot be empty")
    
    min_val = arr[0]
    max_val = arr[0]
    
    for num in arr[1:]:
        if num < min_val:
            min_val = num
        elif num > max_val:
            max_val = num
            
    return min_val, max_val
```

```python
# Test cases including edge cases
assert get_min_max([3, 2, 1, 56, 10000, 167]) == (1, 10000)
assert get_min_max([42]) == (42, 42)                         # Single element
assert get_min_max([5, 5, 5, 5]) == (5, 5)                   # All duplicates
assert get_min_max([-10, -5, -20, -1]) == (-20, -1)          # All negative numbers
assert get_min_max([1, 2, 3, 4, 5]) == (1, 5)                # Sorted ascending

print(" All tests passed!")
```
