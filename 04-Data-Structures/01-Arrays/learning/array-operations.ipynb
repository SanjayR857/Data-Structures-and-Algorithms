# Array Operations & Benchmarks

Experiment with:
- Cost of inserting at index 0 ($O(n)$) vs append at end ($O(1)$ amortized)
- Cost of pop(0) vs pop()
- Dynamic resizing thresholds

```python
# Benchmark placeholder: append vs insert(0)
import time

N = 50000
t0 = time.perf_counter()
lst1 = []
for i in range(N):
    lst1.append(i)
t_append = time.perf_counter() - t0

print(f"Append {N} items: {t_append:.4f} seconds")
```
