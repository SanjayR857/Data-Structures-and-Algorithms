# Array Visualizations & Dry Runs

Use this notebook to visualize pointer movements (e.g., Two Pointers, Dutch National Flag, In-place reversal).

```python
# Two-pointer swap visualization placeholder
def trace_two_pointers(arr):
    left, right = 0, len(arr) - 1
    print(f"Initial: {arr}")
    while left < right:
        arr[left], arr[right] = arr[right], arr[left]
        print(f"Swapped indices ({left}, {right}) -> {arr}")
        left += 1
        right -= 1

trace_two_pointers([1, 2, 3, 4, 5])
```
